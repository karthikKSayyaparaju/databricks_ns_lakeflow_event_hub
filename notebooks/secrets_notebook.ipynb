```python
from databricks.sdk import WorkspaceClient
from getpass import getpass

w = WorkspaceClient()
# w.secrets.create_scope("confluent-kafka-key-value')   
w.secrets.put_secret("confluent-kafka-key-value", "api_key",    string_value=getpass("API key: "))
w.secrets.put_secret("confluent-kafka-key-value", "api_secret", string_value=getpass("API secret: "))
```

```python
api_key    = dbutils.secrets.get(scope="confluent-kafka-key-value", key="api_key")
api_secret = dbutils.secrets.get(scope="confluent-kafka-key-value", key="api_secret")
```

```python
print(" ".join(api_key))
print(" ".join(api_secret))
```
